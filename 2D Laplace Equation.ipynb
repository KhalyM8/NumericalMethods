Automating the Matrix

2 types of blocks, have edges or dont have edges

Check whether matrix is on edge (if i = 0 or nVariable-1, or j = 0 or nVariable-1)

```python
def Create_Matrix_2D_Laplace(nVariables):
    nRows = int(nVariables**(0.5))
    matrix = np.zeros((nVariables,nVariables))
    for i in range(nVariables):
        matrix[i][i] = 4 #Own Tile
        if i < nRows: #hugging top wall
            matrix[i][i+(nRows)] = -1 #Below
            if i % nRows==0: # hugging left wall
                matrix[i][i+1] = -1 #To the right
                continue
            if (i+1) %nRows == 0: #Hugging right wall
                matrix[i][i-1] = -1 #To the left
                continue
            matrix[i][i+1] = -1 #To the right
            matrix[i][i-1] = -1 #To the left
            continue
        if i >= (nRows-1)*nRows: #Hugging Bottom Wall
            matrix[i][i-(nRows)] = -1 #Above
            if i % nRows ==0: # hugging left wall
                matrix[i][i+1] = -1 #To the right
                continue
            if (i+1) %nRows == 0: #Hugging right wall
                matrix[i][i-1] = -1 #To the left
                continue
            matrix[i][i+1] = -1 #To the right
            matrix[i][i-1] = -1 #To the left   
            continue 
        if (i+1) %nRows == 0: #Hugging right wall
            matrix[i][i-1] = -1 #To the left
            matrix[i][i-(nRows)] = -1 #Above
            matrix[i][i+(nRows)] = -1 #Below
            continue
        if i % nRows ==0: # hugging left wall
            matrix[i][i-(nRows)] = -1 #Above
            matrix[i][i+(nRows)] = -1 #Below
            matrix[i][i+1] = -1 #To the right  
            continue
        
        matrix[i][i-1] = -1 #To the left
        matrix[i][i-(nRows)] = -1 #Above
        matrix[i][i+(nRows)] = -1 #Below
        matrix[i][i+1] = -1 #To the right 
            

    
    return matrix
```

```python
boundaryConditions = np.array([20,40,100,120])
A = Create_Matrix_2D_Laplace(4)
np.linalg.inv(A) @ boundaryConditions
```
